```python
import numpy as np
import pandas as pd
import tensorflow as tf
import random
import time
import os
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import *
from tensorflow.keras.optimizers.legacy import Adam

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import *

from scipy.stats import ttest_rel, chi2, binom
from itertools import combinations

# ================= DISABLE XLA =================
tf.config.optimizer.set_jit(False)

# ================= GPU CHECK =================
print("TensorFlow:", tf.__version__)
print("GPUs:", tf.config.list_physical_devices("GPU"))

# ================= CONFIG =================
LR = 0.001
ROUNDS = 30
LOCAL_EPOCHS = 2
BATCH = 16
MU = 0.01
CLIENTS = 4
SEEDS = [42, 3, 7, 72, 82]
MODELS = ["CNN", "LSTM", "BiLSTM", "GRU", "BiGRU"]
CLIENT_TEST_SIZE = 0.2
MODEL_SAVE_DIR = "saved_models"

os.makedirs(MODEL_SAVE_DIR, exist_ok=True)

# ================= LOAD DATA =================
df = pd.read_csv("Dataset(111k).csv")

for col in df.columns:
    if df[col].dtype == object and df[col].astype(str).str.contains(".txt", na=False, regex=False).any():
        df.drop(columns=[col], inplace=True)

df = df.fillna("unknown")

for col in df.columns:
    if df[col].dtype == object:
        df[col] = LabelEncoder().fit_transform(df[col].astype(str))

X = df.drop("label", axis=1).values
y = df["label"].values

X = StandardScaler().fit_transform(X)

print("Dataset shape:", X.shape)
print("Classes:", np.unique(y))

# ================= HELPERS =================
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)

def iid_partition(X, y, K):
    idx = np.arange(len(y))
    clients = [[] for _ in range(K)]

    for c in np.unique(y):
        c_idx = idx[y == c]
        np.random.shuffle(c_idx)
        split = np.array_split(c_idx, K)

        for i in range(K):
            clients[i].extend(split[i])

    output = []

    for c in clients:
        c = np.asarray(c, dtype=int)
        np.random.shuffle(c)
        output.append((X[c], y[c]))

    return output

def format_xy(Xc, model_name):
    return Xc.reshape(len(Xc), -1, 1)

def fedprox_aggregate(global_model, local_models, sizes):
    weights = [m.get_weights() for m in local_models]
    new_weights = [np.average(w, axis=0, weights=sizes) for w in zip(*weights)]
    global_model.set_weights(new_weights)

def model_size(model):
    temp_file = "tmp_model.h5"
    model.save(temp_file, include_optimizer=False)
    size = os.path.getsize(temp_file) / (1024 * 1024)
    os.remove(temp_file)
    return size

def stats(x, metric):
    x = np.asarray(x, dtype=float)
    x = x[~np.isnan(x)]

    if len(x) == 0:
        return np.nan, np.nan, np.nan, np.nan

    mean = np.mean(x)
    sd = np.std(x)
    ci = 1.96 * sd / np.sqrt(len(x))
    low = mean - ci
    high = mean + ci

    if metric in ["Accuracy", "Precision", "Recall", "F1", "AUC", "Specificity"]:
        low = max(0, low)
        high = min(1, high)

    if metric == "MCC":
        low = max(-1, low)
        high = min(1, high)

    return mean, sd, low, high

def mcnemar_test(y_true_a, y_pred_a, y_true_b, y_pred_b):
    y_true_a = np.asarray(y_true_a).flatten()
    y_true_b = np.asarray(y_true_b).flatten()

    assert len(y_true_a) == len(y_true_b), "Test sets must have the same size."
    assert np.array_equal(y_true_a, y_true_b), "Test sets must contain identical instances."

    y_pred_a = np.asarray(y_pred_a).flatten()
    y_pred_b = np.asarray(y_pred_b).flatten()

    correct_a = y_pred_a == y_true_a
    correct_b = y_pred_b == y_true_b

    n01 = int(np.sum(correct_a & ~correct_b))
    n10 = int(np.sum(~correct_a & correct_b))
    n = n01 + n10

    if n == 0:
        return np.nan, 1.0, n01, n10

    if n < 25:
        p = 2 * min(binom.cdf(min(n01, n10), n, 0.5), 0.5)
        p = min(p, 1.0)
        stat = np.nan
    else:
        stat = ((abs(n01 - n10) - 1) ** 2) / n
        p = 1 - chi2.cdf(stat, 1)

    return stat, p, n01, n10

# ================= MODELS =================
def build(name, shape=None, dim=None):
    if name == "CNN":
        model = Sequential([
            Input(shape),
            Conv1D(64, 3, activation="relu"),
            Dropout(0.5),
            Conv1D(32, 3, activation="relu"),
            GlobalMaxPooling1D(),
            Dense(32, activation="relu"),
            Dropout(0.5),
            Dense(1, activation="sigmoid")
        ])

    elif name == "LSTM":
        model = Sequential([
            Input(shape),
            LSTM(128, return_sequences=True),
            Dropout(0.5),
            LSTM(64),
            Dropout(0.5),
            Dense(1, activation="sigmoid")
        ])

    elif name == "BiLSTM":
        model = Sequential([
            Input(shape),
            Bidirectional(LSTM(128, return_sequences=True)),
            Dropout(0.5),
            Bidirectional(LSTM(64)),
            Dropout(0.5),
            Dense(1, activation="sigmoid")
        ])

    elif name == "GRU":
        model = Sequential([
            Input(shape),
            GRU(128, return_sequences=True),
            Dropout(0.5),
            GRU(64),
            Dropout(0.5),
            Dense(1, activation="sigmoid")
        ])

    elif name == "BiGRU":
        model = Sequential([
            Input(shape),
            Bidirectional(GRU(128, return_sequences=True)),
            Dropout(0.5),
            Bidirectional(GRU(64)),
            Dropout(0.5),
            Dense(1, activation="sigmoid")
        ])

    else:
        raise ValueError(f"Unknown model: {name}")

    model.compile(
        optimizer=Adam(learning_rate=LR),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model

# ================= STORAGE =================
rows = []
client_rows = []

results_round = {
    m: {
        "acc": [],
        "loss": [],
        "cm": []
    }
    for m in MODELS
}

y_preds = {}
y_trues = {}

# ================= TRAIN =================
for model_name in MODELS:
    print(f"\n================ MODEL: {model_name} ================")

    for seed in SEEDS:
        print(f"\n------ SEED {seed} ------")

        set_seed(seed)

        # IID partition of complete dataset
        client_full = iid_partition(X, y, CLIENTS)

        # Local train/test split for every client
        client_train = []
        client_test = []

        for cid, (Xc, yc) in enumerate(client_full):
            try:
                Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(
                    Xc,
                    yc,
                    test_size=CLIENT_TEST_SIZE,
                    stratify=yc,
                    random_state=seed
                )
            except ValueError:
                Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(
                    Xc,
                    yc,
                    test_size=CLIENT_TEST_SIZE,
                    random_state=seed
                )

            client_train.append((Xc_tr, yc_tr))
            client_test.append((Xc_te, yc_te))

        # Global test set = union of all client test sets
        X_test_global = np.concatenate([Xc for Xc, _ in client_test], axis=0)
        y_test_global = np.concatenate([yc for _, yc in client_test], axis=0)

        print("Global test size:", len(y_test_global))

        create = lambda: build(model_name, shape=(X.shape[1], 1))
        fmt = lambda Xc: format_xy(Xc, model_name)

        global_model = create()

        acc_hist = []
        loss_hist = []
        comm = 0
        total_train_time = 0.0

        # ================= FEDERATED ROUNDS =================
        for r in range(ROUNDS):
            print(f"\n[Model={model_name} | Seed={seed} | Round={r+1}/{ROUNDS}]")

            local_models = []
            sizes = []

            # ================= CLIENT TRAINING =================
            for cid, (Xc, yc) in enumerate(client_train):
                local_model = create()
                local_model.set_weights(global_model.get_weights())

                global_vars = [
                    tf.identity(v)
                    for v in global_model.trainable_variables
                ]

                Xc_fmt = fmt(Xc)
                yc_float = yc.astype("float32").reshape(-1, 1)

                n = len(yc)
                optimizer = local_model.optimizer
                bce = tf.keras.losses.BinaryCrossentropy()

                t0 = time.time()

                for epoch in range(LOCAL_EPOCHS):
                    permutation = np.random.permutation(n)

                    for start in range(0, n, BATCH):
                        batch_idx = permutation[start:start + BATCH]

                        xb = Xc_fmt[batch_idx]
                        yb = yc_float[batch_idx]

                        with tf.GradientTape() as tape:
                            predictions = local_model(xb, training=True)

                            classification_loss = bce(yb, predictions)

                            prox = tf.add_n([
                                tf.reduce_sum(tf.square(local_var - global_var))
                                for local_var, global_var in zip(
                                    local_model.trainable_variables,
                                    global_vars
                                )
                            ])

                            loss = classification_loss + (MU / 2.0) * prox

                        gradients = tape.gradient(
                            loss,
                            local_model.trainable_variables
                        )

                        gradients_and_variables = [
                            (grad, var)
                            for grad, var in zip(
                                gradients,
                                local_model.trainable_variables
                            )
                            if grad is not None
                        ]

                        optimizer.apply_gradients(gradients_and_variables)

                t1 = time.time()

                total_train_time += t1 - t0

                train_loss, train_acc = local_model.evaluate(
                    Xc_fmt,
                    yc,
                    verbose=0
                )

                print(
                    f" Client {cid} → "
                    f"Loss={train_loss:.4f}, "
                    f"Acc={train_acc:.4f}, "
                    f"Time={t1-t0:.2f}s, "
                    f"N={len(yc)}"
                )

                local_models.append(local_model)
                sizes.append(len(yc))

            # ================= SERVER AGGREGATION =================
            fedprox_aggregate(global_model, local_models, sizes)

            # ================= GLOBAL TEST =================
            global_loss, global_acc = global_model.evaluate(
                fmt(X_test_global),
                y_test_global,
                verbose=0
            )

            print(
                f" >>> GLOBAL TEST → "
                f"Loss={global_loss:.4f}, "
                f"Acc={global_acc:.4f}"
            )

            acc_hist.append(global_acc)
            loss_hist.append(global_loss)

            weight_bytes = sum(w.size for w in global_model.get_weights()) * 4
            comm += weight_bytes * CLIENTS * 2

        # ================= FINAL GLOBAL PREDICTION =================
        t0 = time.time()

        y_prob = global_model.predict(
            fmt(X_test_global),
            verbose=0
        ).flatten()

        inference_time = time.time() - t0

        y_pred = (y_prob > 0.5).astype(int)

        # ================= STORE PREDICTIONS =================
        y_preds[(model_name, seed)] = y_pred.copy()

        if seed not in y_trues:
            y_trues[seed] = y_test_global.flatten().copy()
        else:
            assert np.array_equal(
                y_trues[seed],
                y_test_global.flatten()
            ), f"Test set mismatch for seed {seed}"

        # ================= CONFUSION MATRIX =================
        cm = confusion_matrix(
            y_test_global,
            y_pred,
            labels=[0, 1]
        )

        tn, fp, fn, tp = cm.ravel()

        print("\nFinal Confusion Matrix:")
        print(cm)

        # ================= SAVE MODEL =================
        model_path = os.path.join(
            MODEL_SAVE_DIR,
            f"{model_name}_seed{seed}.keras"
        )

        global_model.save(model_path)

        print("[saved]", model_path)

        # ================= GLOBAL METRICS =================
        rows.append({
            "Model": model_name,
            "Seed": seed,
            "Loss": log_loss(y_test_global, y_prob),
            "Accuracy": accuracy_score(y_test_global, y_pred),
            "Precision": precision_score(y_test_global, y_pred, zero_division=0),
            "Recall": recall_score(y_test_global, y_pred, zero_division=0),
            "F1": f1_score(y_test_global, y_pred, zero_division=0),
            "MCC": matthews_corrcoef(y_test_global, y_pred),
            "AUC": roc_auc_score(y_test_global, y_prob),
            "Specificity": tn / (tn + fp) if (tn + fp) > 0 else np.nan,
            "Training_Time_sec": total_train_time,
            "Inference_Time_sec": inference_time,
            "Model_Size_MB": model_size(global_model),
            "Comm_per_round_MB": comm / (ROUNDS * 1024 * 1024),
            "Total_Comm_MB": comm / (1024 * 1024),
            "Saved_Model_Path": model_path
        })

        results_round[model_name]["acc"].append(acc_hist)
        results_round[model_name]["loss"].append(loss_hist)
        results_round[model_name]["cm"].append(cm)

        # ================= CLIENT TEST METRICS =================
        for cid, (Xc_te, yc_te) in enumerate(client_test):
            yc_prob = global_model.predict(
                fmt(Xc_te),
                verbose=0
            ).flatten()

            yc_pred = (yc_prob > 0.5).astype(int)

            cm_client = confusion_matrix(
                yc_te,
                yc_pred,
                labels=[0, 1]
            )

            tn_c, fp_c, fn_c, tp_c = cm_client.ravel()

            if len(np.unique(yc_te)) > 1:
                auc_value = roc_auc_score(yc_te, yc_prob)
            else:
                auc_value = np.nan

            client_rows.append({
                "Model": model_name,
                "Seed": seed,
                "Client": cid,
                "N_test": len(yc_te),
                "Loss": log_loss(yc_te, yc_prob, labels=[0, 1]),
                "Accuracy": accuracy_score(yc_te, yc_pred),
                "Precision": precision_score(yc_te, yc_pred, zero_division=0),
                "Recall": recall_score(yc_te, yc_pred, zero_division=0),
                "F1": f1_score(yc_te, yc_pred, zero_division=0),
                "MCC": matthews_corrcoef(yc_te, yc_pred),
                "AUC": auc_value,
                "Specificity": tn_c / (tn_c + fp_c) if (tn_c + fp_c) > 0 else np.nan
            })

# ================= CONFUSION MATRICES =================
for model_name in MODELS:
    print(f"\n=== {model_name}: Global Test Confusion Matrices ===")

    for seed, cm in zip(
        SEEDS,
        results_round[model_name]["cm"]
    ):
        print(f"\nSeed {seed}:")
        print(cm)

# ================= AVERAGE OVER SEEDS =================
for model_name in MODELS:
    results_round[model_name]["acc"] = np.mean(
        results_round[model_name]["acc"],
        axis=0
    )

    results_round[model_name]["loss"] = np.mean(
        results_round[model_name]["loss"],
        axis=0
    )

    results_round[model_name]["cm"] = np.mean(
        results_round[model_name]["cm"],
        axis=0
    )

for model_name in MODELS:
    print(f"\n=== {model_name} averaged over {len(SEEDS)} seeds ===")

    print(
        "Confusion Matrix:\n",
        np.round(results_round[model_name]["cm"], 2)
    )

    print(
        "Accuracy per round:",
        np.round(results_round[model_name]["acc"], 4)
    )

    print(
        "Loss per round:",
        np.round(results_round[model_name]["loss"], 4)
    )

# ================= GLOBAL RESULTS DATAFRAME =================
results_df = pd.DataFrame(rows)

summary_metrics = [
    "Loss",
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "MCC",
    "AUC",
    "Specificity",
    "Training_Time_sec",
    "Inference_Time_sec",
    "Model_Size_MB",
    "Comm_per_round_MB",
    "Total_Comm_MB"
]

# ================= PER MODEL SUMMARY =================
per_model_summary = []

for model_name in MODELS:
    sub = results_df[
        results_df["Model"] == model_name
    ]

    for metric in summary_metrics:
        mean, sd, low, high = stats(
            sub[metric],
            metric
        )

        per_model_summary.append({
            "Model": model_name,
            "Metric": metric,
            "Mean ± SD": f"{mean:.4f} ± {sd:.4f}",
            "95% CI": f"[{low:.4f}, {high:.4f}]"
        })

per_model_summary_df = pd.DataFrame(
    per_model_summary
)

print(
    f"\nPER-MODEL SUMMARY "
    f"(mean ± SD and 95% CI over {len(SEEDS)} seeds):\n"
)

print(
    per_model_summary_df.to_string(
        index=False
    )
)

# ================= WIDE SUMMARY =================
per_model_wide = per_model_summary_df.copy()

per_model_wide["Mean [95% CI]"] = per_model_wide.apply(
    lambda r:
        r["Mean ± SD"].split(" ± ")[0]
        + " "
        + r["95% CI"],
    axis=1
)

per_model_wide = per_model_wide.pivot(
    index="Model",
    columns="Metric",
    values="Mean [95% CI]"
)

per_model_wide = per_model_wide[
    summary_metrics
]

print("\nPER-MODEL WIDE SUMMARY:\n")
print(per_model_wide.to_string())

# ================= OVERALL SUMMARY =================
overall_summary = []

for metric in summary_metrics:
    mean, sd, low, high = stats(
        results_df[metric],
        metric
    )

    overall_summary.append({
        "Metric": metric,
        "Mean ± SD": f"{mean:.4f} ± {sd:.4f}",
        "95% CI": f"[{low:.4f}, {high:.4f}]"
    })

overall_summary_df = pd.DataFrame(
    overall_summary
)

print("\nOVERALL SUMMARY:\n")
print(overall_summary_df.to_string(index=False))

# ================= PER CLIENT SUMMARY =================
client_df = pd.DataFrame(
    client_rows
)

client_metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "MCC",
    "AUC",
    "Specificity",
    "Loss"
]

per_client_summary = (
    client_df
    .groupby(["Model", "Client"])[client_metrics]
    .agg(["mean", "std"])
    .round(4)
)

print(
    f"\nPER-CLIENT TEST SUMMARY "
    f"(mean/std over {len(SEEDS)} seeds):\n"
)

print(per_client_summary)

# ================= PAIRED T-TEST =================
paired_test_metrics = [
    "Accuracy",
    "F1",
    "AUC",
    "Loss"
]

pairwise_rows = []

for metric in paired_test_metrics:
    for model_a, model_b in combinations(MODELS, 2):

        a = (
            results_df[
                results_df["Model"] == model_a
            ]
            .sort_values("Seed")[metric]
            .values
        )

        b = (
            results_df[
                results_df["Model"] == model_b
            ]
            .sort_values("Seed")[metric]
            .values
        )

        t_stat, p_value = ttest_rel(a, b)

        pairwise_rows.append({
            "Metric": metric,
            "Model A": model_a,
            "Model B": model_b,
            "Mean A": round(a.mean(), 4),
            "Mean B": round(b.mean(), 4),
            "t-stat": round(t_stat, 4),
            "p-value": round(p_value, 4),
            "Significant (p<0.05)": p_value < 0.05
        })

pairwise_df = pd.DataFrame(
    pairwise_rows
)

print("\nPAIRWISE PAIRED T-TESTS:\n")
print(pairwise_df.to_string(index=False))

# ================= McNEMAR PER SEED =================
mcnemar_per_seed_rows = []

for seed in SEEDS:
    y_true = y_trues[seed]

    for model_a, model_b in combinations(MODELS, 2):
        stat, p, n01, n10 = mcnemar_test(
            y_true,
            y_preds[(model_a, seed)],
            y_true,
            y_preds[(model_b, seed)]
        )

        mcnemar_per_seed_rows.append({
            "Seed": seed,
            "Model A": model_a,
            "Model B": model_b,
            "A right/B wrong": n01,
            "A wrong/B right": n10,
            "stat": round(stat, 4) if not np.isnan(stat) else "exact",
            "p-value": round(p, 4),
            "Significant (p<0.05)": p < 0.05
        })

mcnemar_per_seed_df = pd.DataFrame(
    mcnemar_per_seed_rows
)

print("\nMcNEMAR TEST — PER SEED:\n")
print(mcnemar_per_seed_df.to_string(index=False))

# ================= McNEMAR POOLED =================
mcnemar_pooled_rows = []

for model_a, model_b in combinations(MODELS, 2):
    n01_total = 0
    n10_total = 0

    for seed in SEEDS:
        y_true = y_trues[seed]

        correct_a = (
            y_preds[(model_a, seed)] == y_true
        )

        correct_b = (
            y_preds[(model_b, seed)] == y_true
        )

        n01_total += int(
            np.sum(correct_a & ~correct_b)
        )

        n10_total += int(
            np.sum(~correct_a & correct_b)
        )

    n = n01_total + n10_total

    if n == 0:
        stat = np.nan
        p = 1.0

    elif n < 25:
        p = min(
            2 * min(
                binom.cdf(
                    min(n01_total, n10_total),
                    n,
                    0.5
                ),
                0.5
            ),
            1.0
        )

        stat = np.nan

    else:
        stat = (
            (abs(n01_total - n10_total) - 1) ** 2
        ) / n

        p = 1 - chi2.cdf(
            stat,
            1
        )

    mcnemar_pooled_rows.append({
        "Model A": model_a,
        "Model B": model_b,
        "A right/B wrong (pooled)": n01_total,
        "A wrong/B right (pooled)": n10_total,
        "stat": round(stat, 4) if not np.isnan(stat) else "exact",
        "p-value": round(p, 4),
        "Significant (p<0.05)": p < 0.05
    })

mcnemar_pooled_df = pd.DataFrame(
    mcnemar_pooled_rows
)

print(
    f"\nMcNEMAR TEST — "
    f"POOLED ACROSS {len(SEEDS)} SEEDS:\n"
)

print(
    mcnemar_pooled_df.to_string(
        index=False
    )
)

# ================= SAVE RESULTS =================
results_df.to_csv(
    "global_results_per_seed.csv",
    index=False
)

client_df.to_csv(
    "client_results_per_seed.csv",
    index=False
)

per_model_summary_df.to_csv(
    "per_model_summary.csv",
    index=False
)

per_model_wide.to_csv(
    "per_model_summary_wide.csv"
)

pairwise_df.to_csv(
    "paired_t_tests.csv",
    index=False
)

mcnemar_per_seed_df.to_csv(
    "mcnemar_per_seed.csv",
    index=False
)

mcnemar_pooled_df.to_csv(
    "mcnemar_pooled.csv",
    index=False
)

print("\nAll results saved successfully.")

# ================= PLOT ACCURACY =================
rounds = np.arange(
    1,
    ROUNDS + 1
)

plt.figure(figsize=(10, 6))

for model_name in MODELS:
    plt.plot(
        rounds,
        results_round[model_name]["acc"],
        label=model_name
    )

plt.title(
    "Accuracy vs Communication Rounds (Global Test)"
)

plt.xlabel(
    "Communication Round"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ================= PLOT LOSS =================
plt.figure(figsize=(10, 6))

for model_name in MODELS:
    plt.plot(
        rounds,
        results_round[model_name]["loss"],
        label=model_name
    )

plt.title(
    "Loss vs Communication Rounds (Global Test)"
)

plt.xlabel(
    "Communication Round"
)

plt.ylabel(
    "Loss"
)

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
