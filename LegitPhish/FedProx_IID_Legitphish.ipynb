```python

import os


os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

import gc
import time
import random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Conv1D,
    Dropout,
    GlobalMaxPooling1D,
    Dense,
    LSTM,
    GRU,
    Bidirectional,
)
from tensorflow.keras.optimizers import Adam

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    roc_auc_score,
    confusion_matrix,
    log_loss,
)

from scipy.stats import ttest_rel, chi2, binom
from itertools import combinations


# ============================================================
# GPU CONFIGURATION
# ============================================================

print("=" * 70)
print("TENSORFLOW / GPU CONFIGURATION")
print("=" * 70)
print("TensorFlow version:", tf.__version__)

gpus = tf.config.list_physical_devices("GPU")

if gpus:
    print(f"Physical GPU(s) detected: {len(gpus)}")

    for i, gpu in enumerate(gpus):
        print(f"GPU {i}: {gpu}")
        try:
            # Let TensorFlow grow GPU allocation as required instead of
            # reserving nearly all available GPU memory immediately.
            tf.config.experimental.set_memory_growth(gpu, True)
        except RuntimeError as exc:
            # This only occurs if the GPU runtime was initialized earlier.
            print(f"Could not enable memory growth for GPU {i}: {exc}")

    print("TensorFlow GPU acceleration: ENABLED")
else:
    print("WARNING: No GPU detected. TensorFlow will run on CPU.")

print("=" * 70)


# ============================================================
# CONFIG
# ============================================================

LR = 0.001
ROUNDS = 30
LOCAL_EPOCHS = 2
BATCH = 16
MU = 0.01

CLIENTS = 4
SEEDS = [42, 3, 7, 72, 82]
MODELS = ["CNN", "LSTM", "BiLSTM", "GRU", "BiGRU"]

# Local train/test split ratio inside each client.
CLIENT_TEST_SIZE = 0.20

MODEL_SAVE_DIR = "saved_models"
os.makedirs(MODEL_SAVE_DIR, exist_ok=True)

DATASET_PATH = "url_features_extracted1.csv"


# ============================================================
# LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)
print("\nDataset loaded:", DATASET_PATH)
print("Original dataset shape:", df.shape)

# Remove text/file-name columns containing values such as *.txt.
drop_cols = []
for col in df.columns:
    if df[col].dtype == object:
        if df[col].astype(str).str.contains(r"\.txt", regex=True, na=False).any():
            drop_cols.append(col)

if drop_cols:
    print("Removing text/file columns:", drop_cols)
    df.drop(columns=drop_cols, inplace=True)

# The target must be handled BEFORE filling missing feature values.
# Otherwise, a missing ClassLabel becomes the string "unknown" and is
# subsequently encoded as a third class (e.g., labels {0, 1, 2}).
if "ClassLabel" not in df.columns:
    raise KeyError("The dataset must contain a 'ClassLabel' column.")

# Treat blank target strings as missing as well.
if df["ClassLabel"].dtype == object:
    df["ClassLabel"] = df["ClassLabel"].replace(r"^\s*$", np.nan, regex=True)

missing_target = int(df["ClassLabel"].isna().sum())
if missing_target:
    print(f"Dropping {missing_target} row(s) with missing ClassLabel.")
    df = df.loc[df["ClassLabel"].notna()].copy()

# Fill missing FEATURE values only. This preserves the original use of
# "unknown" for categorical feature values without creating a new target class.
feature_cols = [c for c in df.columns if c != "ClassLabel"]
for col in feature_cols:
    if df[col].dtype == object:
        df[col] = df[col].fillna("unknown")
    else:
        # Keep numeric columns numeric. Median imputation avoids converting an
        # entire numeric feature to an object/string column because of one NaN.
        if df[col].isna().any():
            df[col] = df[col].fillna(df[col].median())

# Encode categorical FEATURE columns.
for col in feature_cols:
    if df[col].dtype == object:
        encoder = LabelEncoder()
        df[col] = encoder.fit_transform(df[col].astype(str))

# Encode the target only when it is categorical/textual.
if df["ClassLabel"].dtype == object:
    target_encoder = LabelEncoder()
    df["ClassLabel"] = target_encoder.fit_transform(df["ClassLabel"].astype(str))
    print("Target mapping:", dict(zip(target_encoder.classes_, target_encoder.transform(target_encoder.classes_))))

X = df.drop("ClassLabel", axis=1).values
y = df["ClassLabel"].to_numpy()

# This script uses sigmoid + binary_crossentropy and therefore requires
# exactly two classes encoded as 0 and 1. Fail early rather than training
# silently with an invalid third label.
unique_labels = np.unique(y)
if len(unique_labels) != 2:
    raise ValueError(
        f"Binary classification requires exactly 2 target classes, but found "
        f"{len(unique_labels)}: {unique_labels.tolist()}"
    )

# If the two numeric labels are not already {0, 1}, map them safely to {0, 1}.
if not np.array_equal(np.sort(unique_labels.astype(float)), np.array([0.0, 1.0])):
    label_map = {label: idx for idx, label in enumerate(sorted(unique_labels.tolist()))}
    print("Remapping binary target labels:", label_map)
    y = np.array([label_map[v] for v in y], dtype=np.int32)

# Preserve the original preprocessing protocol: standardize the complete
# feature matrix before IID client partitioning. Cast to float32 afterward
# so TensorFlow/GPU computation does not unnecessarily use float64.
scaler = StandardScaler()
X = scaler.fit_transform(X).astype(np.float32)
y = y.astype(np.int32)

print("Processed dataset shape:", X.shape)
print("Number of samples:", len(X))
print("Number of features:", X.shape[1])
print("X dtype:", X.dtype)
print("y dtype:", y.dtype)
print("Class distribution:", dict(zip(*np.unique(y, return_counts=True))))

# NOTE:
# There is no upfront global train_test_split.
# For every seed, IID partitioning is applied to the WHOLE dataset and each
# client then performs its own local train/test split.


# ============================================================
# HELPERS
# ============================================================

def set_seed(seed):
    """Set Python, NumPy, and TensorFlow random seeds."""
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)


def iid_partition(X_data, y_data, K):
    """
    IID partitioning.

    Each class is independently shuffled and split approximately evenly
    across K clients, which keeps client class distributions similar.
    """
    idx = np.arange(len(y_data))
    clients = [[] for _ in range(K)]

    for cls in np.unique(y_data):
        cls_idx = idx[y_data == cls].copy()
        np.random.shuffle(cls_idx)
        splits = np.array_split(cls_idx, K)

        for client_id in range(K):
            clients[client_id].extend(splits[client_id])

    client_data = []
    for client_idx in clients:
        client_idx = np.asarray(client_idx, dtype=np.int64)
        np.random.shuffle(client_idx)
        client_data.append(
            (
                X_data[client_idx].astype(np.float32, copy=False),
                y_data[client_idx].astype(np.int32, copy=False),
            )
        )

    return client_data


def format_xy(X_client, model_name):
    """Reshape tabular features to (samples, timesteps/features, channels)."""
    del model_name  # kept in the signature to preserve the original interface
    return X_client.reshape(len(X_client), X_client.shape[1], 1).astype(
        np.float32, copy=False
    )


def fedprox_aggregate(global_model, local_models, sizes):
    """Weighted FedAvg aggregation used after FedProx local optimization."""
    local_weights = [model.get_weights() for model in local_models]

    new_weights = [
        np.average(layer_weights, axis=0, weights=sizes)
        for layer_weights in zip(*local_weights)
    ]

    global_model.set_weights(new_weights)


def model_size(model):
    """Return serialized model size in MB."""
    tmp_path = os.path.join(MODEL_SAVE_DIR, "_tmp_model_size.h5")
    model.save(tmp_path, include_optimizer=False)
    size_mb = os.path.getsize(tmp_path) / (1024 * 1024)
    os.remove(tmp_path)
    return size_mb


def stats(values, metric):
    """Mean, SD, and normal-approximation 95% CI."""
    values = np.asarray(values, dtype=float)
    values = values[~np.isnan(values)]

    if len(values) == 0:
        return np.nan, np.nan, np.nan, np.nan

    mean = np.mean(values)
    sd = np.std(values)
    ci = 1.96 * sd / np.sqrt(len(values))
    low, high = mean - ci, mean + ci

    if metric in ["Accuracy", "Precision", "Recall", "F1", "AUC", "Specificity"]:
        low = max(0, low)
        high = min(1, high)

    if metric == "MCC":
        low = max(-1, low)
        high = min(1, high)

    return mean, sd, low, high


def mcnemar_test(y_true_a, y_pred_a, y_true_b, y_pred_b):
    """
    McNemar's test between two classifiers evaluated on the SAME samples.

    Uses exact binomial testing for fewer than 25 discordant observations and
    continuity-corrected chi-square otherwise.
    """
    y_true_a = np.asarray(y_true_a).flatten()
    y_true_b = np.asarray(y_true_b).flatten()

    assert len(y_true_a) == len(y_true_b), "test sets must be the same size"
    assert np.array_equal(
        y_true_a, y_true_b
    ), "test sets must match (same seed/instances)"

    correct_a = np.asarray(y_pred_a).flatten() == y_true_a
    correct_b = np.asarray(y_pred_b).flatten() == y_true_b

    n01 = int(np.sum(correct_a & ~correct_b))  # A correct, B wrong
    n10 = int(np.sum(~correct_a & correct_b))  # A wrong, B correct
    n = n01 + n10

    if n == 0:
        return np.nan, 1.0, n01, n10

    if n < 25:
        p = 2 * min(binom.cdf(min(n01, n10), n, 0.5), 0.5)
        p = min(p, 1.0)
        stat = np.nan
    else:
        stat = ((abs(n01 - n10) - 1) ** 2) / n
        p = 1 - chi2.cdf(stat, 1)

    return stat, p, n01, n10


# ============================================================
# MODELS
# ============================================================

def build(name, shape=None, dim=None):
    del dim  # preserved for compatibility with the original function signature

    if name == "CNN":
        model = Sequential(
            [
                Input(shape=shape),
                Conv1D(64, 3, activation="relu"),
                Dropout(0.5),
                Conv1D(32, 3, activation="relu"),
                GlobalMaxPooling1D(),
                Dense(32, activation="relu"),
                Dropout(0.5),
                Dense(1, activation="sigmoid"),
            ]
        )

    elif name == "LSTM":
        model = Sequential(
            [
                Input(shape=shape),
                LSTM(128, return_sequences=True),
                Dropout(0.5),
                LSTM(64),
                Dropout(0.5),
                Dense(1, activation="sigmoid"),
            ]
        )

    elif name == "BiLSTM":
        model = Sequential(
            [
                Input(shape=shape),
                Bidirectional(LSTM(128, return_sequences=True)),
                Dropout(0.5),
                Bidirectional(LSTM(64)),
                Dropout(0.5),
                Dense(1, activation="sigmoid"),
            ]
        )

    elif name == "GRU":
        model = Sequential(
            [
                Input(shape=shape),
                GRU(128, return_sequences=True),
                Dropout(0.5),
                GRU(64),
                Dropout(0.5),
                Dense(1, activation="sigmoid"),
            ]
        )

    elif name == "BiGRU":
        model = Sequential(
            [
                Input(shape=shape),
                Bidirectional(GRU(128, return_sequences=True)),
                Dropout(0.5),
                Bidirectional(GRU(64)),
                Dropout(0.5),
                Dense(1, activation="sigmoid"),
            ]
        )

    else:
        raise ValueError(f"Unknown model name: {name}")

    model.compile(
        optimizer=Adam(learning_rate=LR, jit_compile=False),
        loss="binary_crossentropy",
        metrics=["accuracy"],
    )

    return model


# ============================================================
# STORAGE
# ============================================================

# Global union-of-clients test metrics per (model, seed).
rows = []

# Per-client local test metrics per (model, seed, client).
client_rows = []

# Per-round global accuracy/loss and final confusion matrices.
results_round = {m: {"acc": [], "loss": [], "cm": []} for m in MODELS}

# Per-instance predictions for McNemar's test.
y_preds = {}  # (model, seed) -> prediction vector
y_trues = {}  # seed -> aligned global test labels


# ============================================================
# FINAL DEVICE CHECK BEFORE TRAINING
# ============================================================

print("\n" + "=" * 70)
print("DEVICE CHECK BEFORE FEDPROX TRAINING")
print("=" * 70)

logical_gpus = tf.config.list_logical_devices("GPU")

if logical_gpus:
    print("FedProx training can use the following logical GPU device(s):")
    for device in logical_gpus:
        print("  ", device)
else:
    print("WARNING: No logical GPU is available. Training will use CPU.")

print("=" * 70)


# ============================================================
# TRAIN
# ============================================================

for model_name in MODELS:
    print("\n\n" + "#" * 70)
    print(f"MODEL: {model_name}")
    print("#" * 70)

    for seed in SEEDS:
        print("\n" + "=" * 70)
        print(f"MODEL = {model_name} | SEED = {seed}")
        print("=" * 70)

        # Clear Keras state left by the previous completed experiment.
        # This is done only between independent model/seed runs, never inside
        # an active FedProx communication round.
        tf.keras.backend.clear_session()
        gc.collect()

        set_seed(seed)

        # ----------------------------------------------------
        # IID partition of the WHOLE dataset across clients
        # ----------------------------------------------------
        client_full = iid_partition(X, y, CLIENTS)

        # ----------------------------------------------------
        # Each client locally splits its IID shard into train/test
        # ----------------------------------------------------
        client_train = []
        client_test = []

        for cid, (Xc, yc) in enumerate(client_full):
            try:
                Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(
                    Xc,
                    yc,
                    test_size=CLIENT_TEST_SIZE,
                    stratify=yc,
                    random_state=seed,
                )
            except ValueError:
                # Fallback if a shard does not contain enough observations
                # from a class for stratification.
                Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(
                    Xc,
                    yc,
                    test_size=CLIENT_TEST_SIZE,
                    random_state=seed,
                )

            Xc_tr = Xc_tr.astype(np.float32, copy=False)
            Xc_te = Xc_te.astype(np.float32, copy=False)
            yc_tr = yc_tr.astype(np.int32, copy=False)
            yc_te = yc_te.astype(np.int32, copy=False)

            client_train.append((Xc_tr, yc_tr))
            client_test.append((Xc_te, yc_te))

        # ----------------------------------------------------
        # Global/centralized test set = union of local client test shards
        # ----------------------------------------------------
        X_test_global = np.concatenate(
            [Xc for Xc, _ in client_test], axis=0
        ).astype(np.float32, copy=False)
        y_test_global = np.concatenate(
            [yc for _, yc in client_test], axis=0
        ).astype(np.int32, copy=False)

        print("\nClient distributions:")
        for cid, ((Xtr, ytr), (Xte, yte)) in enumerate(
            zip(client_train, client_test)
        ):
            train_classes = dict(zip(*np.unique(ytr, return_counts=True)))
            test_classes = dict(zip(*np.unique(yte, return_counts=True)))

            print(
                f" Client {cid}: "
                f"Train={len(ytr)}, Test={len(yte)}, "
                f"Train classes={train_classes}, Test classes={test_classes}"
            )

        def create_model():
            return build(model_name, shape=(X.shape[1], 1))

        def fmt(Xc):
            return format_xy(Xc, model_name)

        global_model = create_model()

        print("\nGlobal model created")
        print("Model:", model_name)
        print("Input shape:", (X.shape[1], 1))
        print("Trainable parameters:", global_model.count_params())

        acc_hist = []
        loss_hist = []
        comm = 0
        total_train_time = 0.0

        # ====================================================
        # FEDPROX COMMUNICATION ROUNDS
        # ====================================================
        for r in range(ROUNDS):
            print(
                f"\n[Model={model_name} | Seed={seed} | Round={r + 1}/{ROUNDS}]"
            )

            local_models = []
            sizes = []

            # ------------------------------------------------
            # Client local training
            # ------------------------------------------------
            for cid, (Xc, yc) in enumerate(client_train):
                local_model = create_model()
                local_model.set_weights(global_model.get_weights())

                # Copy ONLY global trainable variables. This exactly aligns
                # with local_model.trainable_variables in the proximal term.
                global_vars = [
                    tf.stop_gradient(tf.identity(v))
                    for v in global_model.trainable_variables
                ]

                Xc_fmt = fmt(Xc)
                yc_f = yc.astype(np.float32).reshape(-1, 1)
                n = len(yc)

                optimizer = local_model.optimizer
                bce = tf.keras.losses.BinaryCrossentropy()

                t0 = time.time()

                for epoch in range(LOCAL_EPOCHS):
                    permutation = np.random.permutation(n)

                    for start in range(0, n, BATCH):
                        batch_idx = permutation[start : start + BATCH]
                        xb = Xc_fmt[batch_idx]
                        yb = yc_f[batch_idx]

                        with tf.GradientTape() as tape:
                            predictions = local_model(xb, training=True)
                            base_loss = bce(yb, predictions)

                            # FedProx proximal term:
                            # (MU / 2) * ||w_local - w_global||^2
                            prox = tf.add_n(
                                [
                                    tf.reduce_sum(tf.square(w - gw))
                                    for w, gw in zip(
                                        local_model.trainable_variables,
                                        global_vars,
                                    )
                                ]
                            )

                            loss = base_loss + (MU / 2.0) * prox

                        gradients = tape.gradient(
                            loss, local_model.trainable_variables
                        )
                        optimizer.apply_gradients(
                            zip(gradients, local_model.trainable_variables)
                        )

                t1 = time.time()
                local_elapsed = t1 - t0
                total_train_time += local_elapsed

                train_loss, train_acc = local_model.evaluate(
                    Xc_fmt, yc, verbose=0
                )

                print(
                    f" Client {cid} -> "
                    f"Loss={train_loss:.4f}, "
                    f"Acc={train_acc:.4f}, "
                    f"Time={local_elapsed:.2f}s, "
                    f"N={len(yc)}"
                )

                local_models.append(local_model)
                sizes.append(len(yc))

            # ------------------------------------------------
            # Server aggregation
            # ------------------------------------------------
            fedprox_aggregate(global_model, local_models, sizes)

            # Local models are no longer needed after aggregation.
            # Do not call clear_session() here because global_model is active.
            del local_models
            if "local_model" in locals():
                del local_model
            gc.collect()

            # ------------------------------------------------
            # Global test evaluation after this round
            # ------------------------------------------------
            round_loss, round_acc = global_model.evaluate(
                fmt(X_test_global), y_test_global, verbose=0
            )

            print(
                f" >>> GLOBAL (centralized test) -> "
                f"Loss={round_loss:.4f}, Acc={round_acc:.4f}"
            )

            acc_hist.append(round_acc)
            loss_hist.append(round_loss)

            # Communication cost:
            # float32 = 4 bytes per parameter value.
            # Each round sends global weights down to every client and receives
            # one local model update back from every client.
            weight_bytes = sum(w.size for w in global_model.get_weights()) * 4
            comm += weight_bytes * CLIENTS * 2

        # ====================================================
        # FINAL GLOBAL TEST METRICS
        # ====================================================
        inference_start = time.time()
        y_prob = global_model.predict(fmt(X_test_global), verbose=0).flatten()
        inference_time = time.time() - inference_start

        y_pred = (y_prob > 0.5).astype(np.int32)

        # Save aligned per-instance predictions for McNemar's test.
        y_preds[(model_name, seed)] = y_pred.flatten()

        if seed not in y_trues:
            y_trues[seed] = y_test_global.flatten()
        else:
            assert np.array_equal(
                y_trues[seed], y_test_global.flatten()
            ), (
                f"Test set mismatch for seed {seed}. "
                "McNemar's test requires identical instances across models."
            )

        tn, fp, fn, tp = confusion_matrix(
            y_test_global, y_pred, labels=[0, 1]
        ).ravel()

        # ----------------------------------------------------
        # Save trained global model
        # ----------------------------------------------------
        model_path = os.path.join(
            MODEL_SAVE_DIR, f"{model_name}_seed{seed}.keras"
        )
        global_model.save(model_path)
        print(f" [saved] {model_path}")

        rows.append(
            {
                "Model": model_name,
                "Seed": seed,
                "Loss": log_loss(y_test_global, y_prob, labels=[0, 1]),
                "Accuracy": accuracy_score(y_test_global, y_pred),
                "Precision": precision_score(
                    y_test_global, y_pred, zero_division=0
                ),
                "Recall": recall_score(y_test_global, y_pred, zero_division=0),
                "F1": f1_score(y_test_global, y_pred, zero_division=0),
                "MCC": matthews_corrcoef(y_test_global, y_pred),
                "AUC": roc_auc_score(y_test_global, y_prob),
                "Specificity": tn / (tn + fp) if (tn + fp) > 0 else np.nan,
                "Training_Time_sec": total_train_time,
                "Inference_Time_sec": inference_time,
                "Model_Size_MB": model_size(global_model),
                "Comm_per_round_MB": comm / (ROUNDS * 1024 * 1024),
                "Total_Comm_MB": comm / (1024 * 1024),
                "Saved_Model_Path": model_path,
            }
        )

        results_round[model_name]["acc"].append(acc_hist)
        results_round[model_name]["loss"].append(loss_hist)
        results_round[model_name]["cm"].append(
            confusion_matrix(y_test_global, y_pred, labels=[0, 1])
        )

        # ====================================================
        # PER-CLIENT LOCAL TEST EVALUATION
        # ====================================================
        for cid, (Xc_te, yc_te) in enumerate(client_test):
            yc_prob = global_model.predict(fmt(Xc_te), verbose=0).flatten()
            yc_pred = (yc_prob > 0.5).astype(np.int32)

            cm_c = confusion_matrix(yc_te, yc_pred, labels=[0, 1])
            tn_c, fp_c, fn_c, tp_c = cm_c.ravel()

            auc_val = (
                roc_auc_score(yc_te, yc_prob)
                if len(np.unique(yc_te)) > 1
                else np.nan
            )

            client_rows.append(
                {
                    "Model": model_name,
                    "Seed": seed,
                    "Client": cid,
                    "N_test": len(yc_te),
                    "Loss": log_loss(yc_te, yc_prob, labels=[0, 1]),
                    "Accuracy": accuracy_score(yc_te, yc_pred),
                    "Precision": precision_score(
                        yc_te, yc_pred, zero_division=0
                    ),
                    "Recall": recall_score(yc_te, yc_pred, zero_division=0),
                    "F1": f1_score(yc_te, yc_pred, zero_division=0),
                    "MCC": matthews_corrcoef(yc_te, yc_pred),
                    "AUC": auc_val,
                    "Specificity": (
                        tn_c / (tn_c + fp_c)
                        if (tn_c + fp_c) > 0
                        else np.nan
                    ),
                }
            )

        # Release large objects before the next independent seed run.
        del global_model
        del client_full
        del client_train
        del client_test
        del X_test_global
        del y_test_global
        gc.collect()


# ============================================================
# PRINT ALL CONFUSION MATRICES PER SEED (GLOBAL TEST)
# ============================================================

for model_name in MODELS:
    print(f"\n=== {model_name}: Global Test Confusion Matrices per Seed ===")

    for seed, cm in zip(SEEDS, results_round[model_name]["cm"]):
        print(f"Seed {seed}:\n{cm}")


# ============================================================
# MEAN OVER SEEDS
# Global-test per-round curves + average confusion matrix
# ============================================================

for model_name in MODELS:
    results_round[model_name]["acc"] = np.mean(
        results_round[model_name]["acc"], axis=0
    )
    results_round[model_name]["loss"] = np.mean(
        results_round[model_name]["loss"], axis=0
    )
    results_round[model_name]["cm"] = np.mean(
        results_round[model_name]["cm"], axis=0
    )

for model_name in MODELS:
    print(
        f"\n=== {model_name} "
        f"(Global Test, averaged over {len(SEEDS)} seeds) ==="
    )
    print("Confusion Matrix (avg):\n", np.round(results_round[model_name]["cm"], 2))
    print(
        "Accuracy per round:",
        np.round(results_round[model_name]["acc"], 4),
    )
    print(
        "Loss per round:",
        np.round(results_round[model_name]["loss"], 4),
    )


# ============================================================
# PER-MODEL SUMMARY
# Mean +/- SD and 95% CI over seeds on global test set
# ============================================================

results_df = pd.DataFrame(rows)

summary_metrics = [
    "Loss",
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "MCC",
    "AUC",
    "Specificity",
    "Training_Time_sec",
    "Inference_Time_sec",
    "Model_Size_MB",
    "Comm_per_round_MB",
    "Total_Comm_MB",
]

per_model_summary = []

for model_name in MODELS:
    sub = results_df[results_df["Model"] == model_name]

    for metric in summary_metrics:
        mean, sd, low, high = stats(sub[metric], metric)

        per_model_summary.append(
            {
                "Model": model_name,
                "Metric": metric,
                "Mean +/- SD": f"{mean:.4f} +/- {sd:.4f}",
                "95% CI": f"[{low:.4f}, {high:.4f}]",
            }
        )

per_model_summary_df = pd.DataFrame(per_model_summary)

print(
    f"\nPER-MODEL SUMMARY "
    f"(mean +/- SD, 95% CI over {len(SEEDS)} seeds, Global Test Set):\n"
)
print(per_model_summary_df.to_string(index=False))

# Wide pivot: one row per model, one column per metric.
per_model_wide = per_model_summary_df.copy()
per_model_wide["Mean [95% CI]"] = per_model_wide.apply(
    lambda row: row["Mean +/- SD"].split(" +/- ")[0] + " " + row["95% CI"],
    axis=1,
)

per_model_wide = per_model_wide.pivot(
    index="Model", columns="Metric", values="Mean [95% CI]"
)
per_model_wide = per_model_wide[summary_metrics]

print(
    f"\nPER-MODEL SUMMARY, WIDE "
    f"(mean [95% CI] over {len(SEEDS)} seeds, Global Test Set):\n"
)
print(per_model_wide.to_string())


# ============================================================
# OVERALL POOLED SUMMARY
# ============================================================

overall_summary = []

for metric in summary_metrics:
    mean, sd, low, high = stats(results_df[metric], metric)

    overall_summary.append(
        {
            "Metric": metric,
            "Mean +/- SD": f"{mean:.4f} +/- {sd:.4f}",
            "95% CI": f"[{low:.4f}, {high:.4f}]",
        }
    )

overall_summary_df = pd.DataFrame(overall_summary)

print(
    f"\nSAVED MODELS ({len(results_df)} total, in '{MODEL_SAVE_DIR}/'):\n"
)
print(results_df[["Model", "Seed", "Saved_Model_Path"]].to_string(index=False))

print("\nOVERALL SUMMARY (all models & seeds pooled):\n")
print(overall_summary_df.to_string(index=False))


# ============================================================
# PER-CLIENT TEST SUMMARY
# Averaged over seeds
# ============================================================

client_df = pd.DataFrame(client_rows)

client_metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "MCC",
    "AUC",
    "Specificity",
    "Loss",
]

per_client_summary = (
    client_df.groupby(["Model", "Client"])[client_metrics]
    .agg(["mean", "std"])
    .round(4)
)

print(
    f"\nPER-CLIENT TEST SUMMARY "
    f"(mean/std over {len(SEEDS)} seeds):\n"
)
print(per_client_summary)


# ============================================================
# PAIRWISE PAIRED T-TESTS ACROSS MODELS
# Same seed-level global test results
# ============================================================

paired_test_metrics = ["Accuracy", "F1", "AUC", "Loss"]
pairwise_rows = []

for metric in paired_test_metrics:
    for model_a, model_b in combinations(MODELS, 2):
        a = (
            results_df[results_df["Model"] == model_a]
            .sort_values("Seed")[metric]
            .values
        )
        b = (
            results_df[results_df["Model"] == model_b]
            .sort_values("Seed")[metric]
            .values
        )

        t_stat, p_val = ttest_rel(a, b)

        pairwise_rows.append(
            {
                "Metric": metric,
                "Model A": model_a,
                "Model B": model_b,
                "Mean A": round(a.mean(), 4),
                "Mean B": round(b.mean(), 4),
                "t-stat": round(t_stat, 4),
                "p-value": round(p_val, 4),
                "Significant (p<0.05)": p_val < 0.05,
            }
        )

pairwise_df = pd.DataFrame(pairwise_rows)

print(
    f"\nPAIRWISE PAIRED T-TESTS across {len(SEEDS)} seeds "
    f"(Global Test Set):\n"
)
print(pairwise_df.to_string(index=False))


# ============================================================
# McNEMAR'S TEST - PER SEED
# ============================================================

mcnemar_per_seed_rows = []

for seed in SEEDS:
    y_true = y_trues[seed]

    for model_a, model_b in combinations(MODELS, 2):
        stat, p, n01, n10 = mcnemar_test(
            y_true,
            y_preds[(model_a, seed)],
            y_true,
            y_preds[(model_b, seed)],
        )

        mcnemar_per_seed_rows.append(
            {
                "Seed": seed,
                "Model A": model_a,
                "Model B": model_b,
                "A right/B wrong": n01,
                "A wrong/B right": n10,
                "stat": round(stat, 4) if not np.isnan(stat) else "exact",
                "p-value": round(p, 4),
                "Significant (p<0.05)": p < 0.05,
            }
        )

mcnemar_per_seed_df = pd.DataFrame(mcnemar_per_seed_rows)

print("\nMcNEMAR'S TEST - PER SEED (instance-level, Global Test Set):\n")
print(mcnemar_per_seed_df.to_string(index=False))


# ============================================================
# McNEMAR'S TEST - POOLED DISCORDANT COUNTS ACROSS SEEDS
# ============================================================

mcnemar_pooled_rows = []

for model_a, model_b in combinations(MODELS, 2):
    n01_total = 0
    n10_total = 0

    for seed in SEEDS:
        y_true = y_trues[seed]

        correct_a = y_preds[(model_a, seed)] == y_true
        correct_b = y_preds[(model_b, seed)] == y_true

        n01_total += int(np.sum(correct_a & ~correct_b))
        n10_total += int(np.sum(~correct_a & correct_b))

    n = n01_total + n10_total

    if n == 0:
        stat = np.nan
        p = 1.0
    elif n < 25:
        p = min(
            2 * min(binom.cdf(min(n01_total, n10_total), n, 0.5), 0.5),
            1.0,
        )
        stat = np.nan
    else:
        stat = ((abs(n01_total - n10_total) - 1) ** 2) / n
        p = 1 - chi2.cdf(stat, 1)

    mcnemar_pooled_rows.append(
        {
            "Model A": model_a,
            "Model B": model_b,
            "A right/B wrong (pooled)": n01_total,
            "A wrong/B right (pooled)": n10_total,
            "stat": round(stat, 4) if not np.isnan(stat) else "exact",
            "p-value": round(p, 4),
            "Significant (p<0.05)": p < 0.05,
        }
    )

mcnemar_pooled_df = pd.DataFrame(mcnemar_pooled_rows)

print(
    f"\nMcNEMAR'S TEST - POOLED ACROSS {len(SEEDS)} SEEDS "
    f"(Global Test Set):\n"
)
print(mcnemar_pooled_df.to_string(index=False))


# ============================================================
# OPTIONAL CSV OUTPUTS
# ============================================================

results_df.to_csv("fedprox_global_results.csv", index=False)
client_df.to_csv("fedprox_client_results.csv", index=False)
per_model_summary_df.to_csv("fedprox_model_summary.csv", index=False)
pairwise_df.to_csv("fedprox_pairwise_ttests.csv", index=False)
mcnemar_per_seed_df.to_csv("fedprox_mcnemar_per_seed.csv", index=False)
mcnemar_pooled_df.to_csv("fedprox_mcnemar_pooled.csv", index=False)

print("\nSaved result tables:")
print("  fedprox_global_results.csv")
print("  fedprox_client_results.csv")
print("  fedprox_model_summary.csv")
print("  fedprox_pairwise_ttests.csv")
print("  fedprox_mcnemar_per_seed.csv")
print("  fedprox_mcnemar_pooled.csv")


# ============================================================
# PLOTS
# ============================================================

round_numbers = np.arange(1, ROUNDS + 1)

plt.figure()
for model_name in MODELS:
    plt.plot(
        round_numbers,
        results_round[model_name]["acc"],
        label=model_name,
    )

plt.title("Accuracy vs Rounds (Global Test)")
plt.xlabel("Communication Round")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()

plt.figure()
for model_name in MODELS:
    plt.plot(
        round_numbers,
        results_round[model_name]["loss"],
        label=model_name,
    )

plt.title("Loss vs Rounds (Global Test)")
plt.xlabel("Communication Round")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()

print("\nFedProx experiment completed successfully.")
```
